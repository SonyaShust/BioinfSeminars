```python
n = []
while True:
    num = int(input("Вводите числа, введите 0 для завершения:"))
    if num == 0:
        break
    n.append(num)
otr=False
for i in range(len(n)):
    if n[i] < 0:
        otr=True
        break
if otr:
    for i in range(len(n)):
        n[i]=-n[i]
print(n)
```
