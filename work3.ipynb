```python
n = []
while True:
    num = int(input("Вводите числа, введите 0 для завершения:"))
    if num == 0:
        break
    n.append(num)
otr=False
for i in range(len(n)):
    if n[i] < 0:
        otr=True
        break
if otr:
    for i in range(len(n)):
        n[i]=-n[i]
print(n)
```

```python
s=[1,2,3,4,5,6]
odd = 0
even = 0
for i, value in enumerate(s):
    if i % 2 == 0:
        even += value
    else:
        odd += value
print(even / odd)
```

```python
word = input ("Введите слово")
russ_alphabet = {"А":1, "В":1, "Е":1, "И":1, "Н":1, "О":1, 
                "Р":1, "С":1, "Т":1, "Д":2, "К":2, "Л":2, "М":2,
                "П":2, "У":2, "Б":3, "Г":3, "Ё":3, "Ь":3, "Я":3, 
                "Й":4, "Ы":4, "Ж":5, "З":5, "Х":5, "Ц":5,
                "Ч":5, "Ш":8, "Э":8, "Ю":8, "Ф":10, "Щ":10, "Ъ":10}
eng_alphabet = {"A":1, 'E':1, 'I':1, 'O':1, 'U':1, 'L':1, 'N':1, 
                'S':1, 'T':1, 'R':1, 'D':2, 'G':2, 'B':3, 
                'C':3, 'M':3, 'P':3, 'F':4, 'H':4, 'V':4, 
                'W':4, 'Y':4, 'K':5, 'J':8, 'X':8, 'Q':10, 'Z':10}
first = word [0].upper()
if first in russ_alphabet:
    alp = russ_alphabet
elif first in eng_alphabet:
    alp = eng_alphabet

total = 0
for letter in word.upper():
    if letter in alp:
        total += alp[letter]
    else:
        print('error')
print(total)
```

```python
cats = [('Rachel' , '4' , 'Paola' , 'Lindberg'),
        ('Iris' , '2' , 'Vince' , 'Gree'),
        ('Grysya' , '1' , 'Irina' , 'Roar'),
        ('Momo' , '4' , 'Adele' , 'Beck'),
        ('Chuchu' , '7' , 'Adele' , 'Beck')]
ownerdict = {}
for cat_name, age, first_name, last_name in cats:
    owner=(first_name, last_name)
    catinfo=f'{cat_name},{age}'
    if owner in ownerdict:
        ownerdict[owner].append(catinfo)
    else:
        ownerdict[owner] = [catinfo]
for (first_name, last_name), catslist in ownerdict.items():
    catsresult = ' '.join(catslist)
    print(f"{first_name} {last_name}: {catsresult}")
```
