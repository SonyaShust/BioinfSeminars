```python
cats = [('Rachel' , '4' , 'Paola' , 'Lindberg'),
        ('Iris' , '2' , 'Vince' , 'Gree'),
        ('Grysya' , '1' , 'Irina' , 'Roar'),
        ('Momo' , '4' , 'Adele' , 'Beck'),
        ('Chuchu' , '7' , 'Adele' , 'Beck')]
ownerdict = {}
for cat_name, age, first_name, last_name in cats:
    owner=(first_name, last_name)
    catinfo=f'{cat_name},{age}'
    if owner in ownerdict:
        ownerdict[owner].append(catinfo)
    else:
        ownerdict[owner] = [catinfo]
for (first_name, last_name), catslist in ownerdict.items():
    catsresult = '; '.join(catslist)
    print(f"{first_name} {last_name}: {catsresult}")
```

```python
from random import randint
N = 10
a = []
for i in range (N):
    a.append(randint (1,101))
print (a)
for i in range (N-1):
    swapped = False
    for j in range (N-1-i):
        if a[j] > a[j+1]:
            a[j], a[j+1] = a[j+1], a[j]
            swapped = True
        if not swapped:
            break
print (a)
```

```python
import random
import time

size = [10, 100, 1000, 10000, 100000]
times = []

for s in size:
    test = [random.randint(1, 1000000) for _ in range(s)]
    n = len(test)
    tc = test.copy()
    
    start = time.perf_counter()
    
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if tc[j] > tc[j + 1]:
                tc[j], tc[j + 1] = tc[j + 1], tc[j]
                swapped = True
        if not swapped:
            break
    end = time.perf_counter()
    answer = end - start
    times.append(answer)
    
    print(f"{s}  {answer}")
```

```python
import random
import time
import sys

print("1. У вас есть 3 колышка (1, 2, 3)")
print("2. Цифры означают размер: 1 - самый маленький, 7 - самый большой")
print("3. За один ход можно переместить ВЕРХНИЙ диск с одного колышка на другой")
print("4. НЕЛЬЗЯ класть больший диск на меньший")
print("5. Цель: собрать все диски на ОДНОМ колышке в правильном порядке (большие внизу, маленькие сверху)")
input("\nНажмите Enter, чтобы начать игру...")

disks = list(range(1, 8))
random.shuffle(disks)

rods = [[], [], []]


for disk in disks:
    rod_index = random.randint(0, 2)
    rods[rod_index].append(disk)

for rod in rods:
    rod.sort(reverse=True)
moves_count = 0

while True:
    max_height = 0
    for rod in rods:
        if len(rod) > max_height:
            max_height = len(rod)
    
    
    for level in range(max_height - 1, -1, -1):
        line = ""
        for rod in rods:
            if level < len(rod):
                disk = rod[level]
                line += f"  [{disk:^3}]  "
            else:
                line += "   |    "
        print(line)
    
    print("  (1)     (2)     (3)")
 
    

    win = False
    for rod in rods:
        if len(rod) == 7:
            correct = True
            for i in range(len(rod) - 1):
                if rod[i] <= rod[i + 1]:
                    correct = False
                    break
            if correct:
                win = True
                break
    
    if win:
        print(f"Вы собрали башню за {moves_count} ходов!")
        break
    
    print(f"\nХодов сделано: {moves_count}")
    
    try:
        from_input = input("С какого колышка переместить? (1-3, q - выход): ")
        
        if from_input.lower() == 'q':
            print("\nСпасибо за игру!")
            sys.exit()
        
        from_rod = int(from_input) - 1  # переводим в индекс 0,1,2
        
        to_input = input("На какой колышек переместить? (1-3): ")
        to_rod = int(to_input) - 1
        
    except ValueError:
        print("\n❌ Ошибка: введите число от 1 до 3!")
        input("Нажмите Enter, чтобы продолжить...")
        continue
    
    # ========== ПРОВЕРКА КОРРЕКТНОСТИ ХОДА ==========
    # Проверка, что колышки существуют
    if from_rod not in [0, 1, 2] or to_rod not in [0, 1, 2]:
        print("\n❌ Неверный номер колышка! Используйте 1, 2 или 3.")
        input("Нажмите Enter, чтобы продолжить...")
        continue
    
    # Проверка, что на исходном колышке есть диски
    if not rods[from_rod]:
        print(f"\n❌ На колышке {from_rod + 1} нет дисков!")
        input("Нажмите Enter, чтобы продолжить...")
        continue
    
    # Если целевой колышек не пуст, проверяем правило "больший на меньший нельзя"
    if rods[to_rod]:
        top_from = rods[from_rod][-1]  # верхний диск исходного
        top_to = rods[to_rod][-1]      # верхний диск целевого
        
        if top_from > top_to:
            print(f"\n❌ Нельзя переместить диск {top_from} на диск {top_to} (больший на меньший)!")
            input("Нажмите Enter, чтобы продолжить...")
            continue
    
    # ========== ВЫПОЛНЕНИЕ ХОДА ==========
    disk = rods[from_rod].pop()  # снимаем верхний диск
    rods[to_rod].append(disk)     # кладём на целевой колышек
    moves_count += 1
    
    print("\n✓ Ход выполнен!")
    time.sleep(0.5)
```
