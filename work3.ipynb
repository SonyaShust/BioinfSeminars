```python
cats = [('Rachel' , '4' , 'Paola' , 'Lindberg'),
        ('Iris' , '2' , 'Vince' , 'Gree'),
        ('Grysya' , '1' , 'Irina' , 'Roar'),
        ('Momo' , '4' , 'Adele' , 'Beck'),
        ('Chuchu' , '7' , 'Adele' , 'Beck')]
ownerdict = {}
for cat_name, age, first_name, last_name in cats:
    owner=(first_name, last_name)
    catinfo=f'{cat_name},{age}'
    if owner in ownerdict:
        ownerdict[owner].append(catinfo)
    else:
        ownerdict[owner] = [catinfo]
for (first_name, last_name), catslist in ownerdict.items():
    catsresult = '; '.join(catslist)
    print(f"{first_name} {last_name}: {catsresult}")
```

```python
from random import randint
N = 10
a = []
for i in range (N):
    a.append(randint (1,100))
print (a)
for i in range (N-1):
    swapped = False
    for j in range (N-1-i):
        if a[j] > a[j+1]:
            a[j], a[j+1] = a[j+1], a[j]
            swapped = True
        if not swapped:
            break
print (a)
```

```python
import random
import time

size = [10, 100, 1000, 10000, 100000]
times = []

for s in size:
    test = [random.randint(1, 1000000) for _ in range(s)]
    n = len(test)
    tc = test.copy()
    
    start = time.perf_counter()
    
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if tc[j] > tc[j + 1]:
                tc[j], tc[j + 1] = tc[j + 1], tc[j]
                swapped = True
        if not swapped:
            break
    end = time.perf_counter()
    answer = end - start
   # times.append(answer)
    
    print(f"{s}  {answer}")
```

```python
import random

print("1. У вас есть 3 колышка (1, 2, 3)")
print("2. Цифры означают размер: 1 - самый маленький, 7 - самый большой")
print("3. За один ход можно переместить верхний диск с одного колышка на другой")
print("4. нельзя класть больший диск на меньший")
print("5. Цель: собрать все диски на одном колышке в правильном порядке (большие внизу, маленькие сверху)")
input("Нажмите Enter, чтобы начать.")

disks = [1,2,3,4,5,6,7]
random.shuffle(disks)

rods = [[], [], []]

for disk in disks:
    rod_index = random.randint(0, 2)
    rods[rod_index].append(disk)

for rod in rods:
    rod.sort(reverse=True)

moves = 0

while True:
    print(f'Первый колышек: {rods[0]}')
    print(f'Второй колышек:{rods[1]}')
    print(f'Третий колышек:{rods[2]}')
 
    win = False
    for rod in rods:
        if len(rod) == 7:
            correct = True
            for i in range(len(rod) - 1):
                if rod[i] <= rod[i + 1]:
                    correct = False
                    break
            if correct:
                win = True
                break

    if win:
        print(f"Вы собрали башню за {moves} ходов!")
        break
    
    print(f"Ходов сделано: {moves}")
    
    from_input = input("С какого колышка переместить? (1-3, q - выход): ")
    
    if from_input == 'q':
        print("Спасибо за игру!")
        break
    
    from_rod = int(from_input) - 1
    
    to_input = input("На какой колышек переместить? (1-3): ")
    to_rod = int(to_input) - 1

    # Проверка правила
    if rods[to_rod]:
        fromtop = rods[from_rod]
        totop = rods[to_rod]   
        
        if fromtop > totop:
            print(f"Нельзя переместить диск {fromtop} на диск {totop} так как он больше")
            input("Нажмите Enter, чтобы продолжить...")
            continue
    
    # пользователь делает ход 
    disk = rods[from_rod].pop() 
    rods[to_rod].append(disk)  
    moves += 1
    
    print("Ход выполнен")
```
