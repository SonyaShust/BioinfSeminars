```python
v = input ("Введите числа через пробелы")
a, b =  map(float, v. split())
n = input ("Введите символ операции")
if n == "+":
    print (a + b)
elif n == "-":
    print (a - b)
elif n == "*":
    print(a * b)
elif n == "/" :
    print (a / b)
```

```python
x = int (input("Введите первоначальную сумму")) 
p = int (input("Введите проценты в виде %"))
y = int (input("Введите сумму, не меньше которой должен быть после n лет"))
n= 0
while True:
 if (x * ((1 + (p / 100)) ** n)) >= y :
     print ("n лет:", n)
     break
 n +=1
```

```python
n = int(input("Введите числo:"))
res = 1
for i in range (1, n + 1):
    res = res * i
print(res)
```

```python
n = int(input("Введите № числa:"))
if n<=0:
    print ("Error")
elif n==1 or n==2:
    res=1
else:
    p1 = 1
    p2 = 1
    for i in range (3, n+1):
       c = p1 + p2
       p2 = p1
       p1 = c
    res = p1
print (res)
```
